```python
import pandas as pd
df = pd.DataFrame({
    "math" : [95,96,99,93,96],
    "science" : [91,81,94,90,97],
    "english" : [90,82,86,99,92]
})
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.heatmap(df)
plt.show()
```

```python
sns.heatmap(df,annot=True)
plt.show()
```

```python

sns.heatmap(df,annot=True,cmap="coolwarm")
plt.show()
```

```python
# adding grid lines
sns.heatmap(df,annot=True,linewidth =1,linecolor = 'blue')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(df,annot=True,cmap ="YlGnBu")
plt.title("students marks heatmap")
plt.show()
```
