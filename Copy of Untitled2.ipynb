```python
tips_df = sns.load_dataset('tips')
```

```python
tips_df
```

```python
sns.lineplot(x = 'total_bill',y = "tip",data = tips_df)
```

```python
sns.lineplot(x = 'size',y = "smoker",data = tips_df)
```

```python
sns.lineplot(x = 'total_bill',y = "time",data = tips_df)
```

```python
sns.lineplot(x = 'size',
            y = 'total_bill',
             data = tips_df,
             hue = 'sex',
             style = 'sex',
             palette = "hot",
             )
```

```python
sns.lineplot(x = 'size',
            y = 'total_bill',
             data = tips_df,
             hue = 'sex',
             style = 'sex',
             palette = "hot",
             dashes = False
             )
```

```python
sns.lineplot(x = 'size',
            y = 'total_bill',
             data = tips_df,
             hue = 'sex',
             style = 'sex',
             palette = "hot",
             dashes = False,
             markers = ['o','<'],
             legend = 'brief'
             )
```

```python
import matplotlib.pyplot as plt
sns.lineplot(x = 'size',
            y = 'total_bill',
             data = tips_df,
             hue = 'sex',
             style = 'sex',
             palette = "hot",
             dashes = False,
             markers = ['o','<'],
             legend = 'brief'
             )
plt.title('Line Plot',fontsize = 19)
plt.xlabel('Size',fontsize = 15)
plt.ylabel('Total Bill',fontsize = 15)
plt.show()
```
